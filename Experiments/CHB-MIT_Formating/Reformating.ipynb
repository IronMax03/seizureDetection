Any given subject should have a dimention of (N, 23, 256) where N is the number of samples, 23 is the number of channels and 256 is the number of time points.

```python
import numpy as np
import pandas as pd

X = np.load("X.npy")
Y = np.load("Y.npy")

print(X.shape)
print(Y.shape)
```

```python
import matplotlib.pyplot as plt

df = pd.DataFrame(Y)

print(df.where(df == 1).count())
print(df.describe())

counts = df.value_counts()
counts.plot(kind="pie", autopct="%1.1f%%", legend=False, title="Seizure vs Non-Seizure Distribution")
plt.show()
```
