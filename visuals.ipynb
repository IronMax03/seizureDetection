```python
import mne
import numpy as np
import matplotlib.pyplot as plt

PATH = "siena-scalp-eeg-database-1.0.0"
raw = mne.io.read_raw_edf(f"{PATH}/PN05/PN05-2.edf", preload=True)
raw.rename_channels(lambda x: x.replace("EEG ", "").replace("-REF", "").strip())

# 1. Drop / mark non-EEG channels so they aren't placed on the scalp
non_eeg = ["EKG EKG", "SPO2", "HR", "1", "2", "MK"]
raw.drop_channels([ch for ch in non_eeg if ch in raw.ch_names])

# 2. Normalize names to the montage's expected casing
montage = mne.channels.make_standard_montage("standard_1020")
montage_names = {name.upper(): name for name in montage.ch_names}
raw.rename_channels(
    {ch: montage_names[ch.upper()] for ch in raw.ch_names
     if ch.upper() in montage_names}
)

raw.set_montage(montage, on_missing="warn")

# 3. Now only EEG channels with real positions remain — pick them explicitly
picks = mne.pick_types(raw.info, eeg=True)
data = raw.get_data(picks=picks)
values = data.mean(axis=1)

fig, ax = plt.subplots()
mne.viz.plot_topomap(values, mne.pick_info(raw.info, picks), axes=ax, show=True)
```

## plot 2

```python
import mne
import numpy as np
from pathlib import Path
from IPython.display import Image

# ---- 3D backend: off-screen rendering (avoids ipywidgets/notebook crashes) ----
import pyvista
pyvista.OFF_SCREEN = True
mne.viz.set_3d_backend("pyvistaqt")

# normalize casing to match the 10-20 montage (Fc1 -> FC1, etc.)
montage = mne.channels.make_standard_montage("standard_1020")
montage_names = {name.upper(): name for name in montage.ch_names}
raw.rename_channels(
    {ch: montage_names[ch.upper()] for ch in raw.ch_names
     if ch.upper() in montage_names}
)
raw.set_montage(montage, on_missing="warn")

# keep only EEG channels that now have positions
raw.pick("eeg")

# ============================================================
# 2. Band power per channel (alpha 8-12 Hz)
# ============================================================
spectrum = raw.compute_psd(fmin=1, fmax=40)
psds, freqs = spectrum.get_data(return_freqs=True)
alpha = (freqs >= 8) & (freqs <= 12)
alpha_power = psds[:, alpha].mean(axis=1)

info = raw.info
evoked = mne.EvokedArray(alpha_power[:, np.newaxis], info)

# ============================================================
# 3. fsaverage template + forward model
# ============================================================
fs_dir = mne.datasets.fetch_fsaverage()
subjects_dir = fs_dir.parent
subject = "fsaverage"

raw.set_eeg_reference("average", projection=True)
evoked.set_eeg_reference("average", projection=True)

fwd = mne.make_forward_solution(
    evoked.info,
    trans="fsaverage",
    src=fs_dir / "bem" / "fsaverage-ico-5-src.fif",
    bem=fs_dir / "bem" / "fsaverage-5120-5120-5120-bem-sol.fif",
    eeg=True,
)

# ============================================================
# 4. Inverse solution -> source estimate
# ============================================================
cov = mne.make_ad_hoc_cov(evoked.info)
inv = mne.minimum_norm.make_inverse_operator(evoked.info, fwd, cov)
stc = mne.minimum_norm.apply_inverse(
    evoked, inv, method="dSPM", lambda2=1.0 / 9.0
)

# ============================================================
# 5. Render 3D cortical map off-screen, save + display
# ============================================================
brain = stc.plot(
    subject=subject,
    subjects_dir=subjects_dir,
    hemi="both",
    surface="inflated",
    time_viewer=False,
    colorbar=True,
)
brain.save_image("source_map.png")
brain.close()

Image("source_map.png")
```

## Gif Plot

```python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
import matplotlib as mpl
import mne

PATH = "siena-scalp-eeg-database-1.0.0"
raw = mne.io.read_raw_edf(f"{PATH}/PN00/PN00-3.edf", preload=True)
raw.rename_channels(lambda x: x.replace("EEG ", "").replace("-REF", "").strip())

# 1. Drop non-EEG channels
non_eeg = ["EKG EKG", "SPO2", "HR", "1", "2", "MK"]
raw.drop_channels([ch for ch in non_eeg if ch in raw.ch_names])

# 2. Normalize names to montage casing
montage = mne.channels.make_standard_montage("standard_1020")
montage_names = {name.upper(): name for name in montage.ch_names}
raw.rename_channels(
    {ch: montage_names[ch.upper()] for ch in raw.ch_names
     if ch.upper() in montage_names}
)
raw.set_montage(montage, on_missing="warn")

# 3. Pick EEG channels (define picks here — it was missing before)
picks = mne.pick_types(raw.info, eeg=True)
info = mne.pick_info(raw.info, picks)
sfreq = raw.info["sfreq"]

# 4. Window + downsample
t_start, t_stop = 60.0, 80.0
start, stop = raw.time_as_index([t_start, t_stop])
seg = raw.get_data(picks=picks)[:, start:stop]
step = int(sfreq // 10)
frames = range(0, seg.shape[1], step)

vmin, vmax = seg.min(), seg.max()

# 5. Figure with room for a colorbar
fig, ax = plt.subplots(figsize=(6, 5.5))
fig.suptitle("PN00-3 — Scalp EEG Topography", fontsize=13, fontweight="bold")

# Persistent colorbar (legend), drawn once from a fixed-scale mappable
norm = mpl.colors.Normalize(vmin=vmin, vmax=vmax)
sm = mpl.cm.ScalarMappable(norm=norm, cmap="RdBu_r")
cbar = fig.colorbar(sm, ax=ax, shrink=0.7, pad=0.05)
cbar.set_label("Voltage (µV)", rotation=270, labelpad=15)

def update(i):
    ax.clear()
    mne.viz.plot_topomap(
        seg[:, i], info, axes=ax, show=False,
        vlim=(vmin, vmax), contours=0, cmap="RdBu_r",
    )
    ax.set_title(f"t = {t_start + i / sfreq:.2f} s", fontsize=10)

ani = FuncAnimation(fig, update, frames=frames, interval=100)
ani.save("seizure_topomap.gif", writer="pillow", fps=10)
plt.close(fig)
```
